```python
from prophet import Prophet
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the EV sales dataset
df = pd.read_csv("simulated_ev_sales.csv", parse_dates=["Date"])
df_monthly = df.groupby("Date")[["EV_Sales"]].sum().reset_index()
df_monthly.columns = ["ds", "y"]

model = Prophet()
model.fit(df_monthly)

future = model.make_future_dataframe(periods=72, freq="M")
forecast = model.predict(future)

fig = model.plot(forecast)
plt.title("Global EV Sales Forecast Through 2030")
plt.xlabel("Date")
plt.ylabel("EV Sales")
plt.show()
```

```python
#1. Which countries lead in EV adoption? We'll analyze the dataset to find which countries have the highest EV sales in total. 
#This will allow us to identify the leading countries in EV adoption.

df.groupby("Country")["EV_Sales"].sum().sort_values(ascending=False)
```

```python
#2. How do government incentives impact sales? 
#We will analyze the relationship between government incentives and EV sales to determine if there's a correlation between the two.
df[["EV_Sales", "Incentives"]].corr()
```
